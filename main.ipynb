```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

print("1. Load CSV File")

try:
    path = input("Enter Dataset Path (e.g. data/sales_data.csv): ").strip()
    data = pd.read_csv(path)
    print("Dataset Loaded Successfully!")
 
except :
    print("Invalid file name")
```

```python
print("========== EXPLORE DATASET ==========")
print("1. Display First 5 Rows")
print("2. Display Last 5 Rows")
print("3. Display Column Names")
print("4. Display Data Types")
print("5. Display Dataset Information")
print("6. Display Statistical Summary")
 
try:
    choice = input("Enter Your Choice: ")

    if choice == '1'  :
        print(data.head())

    elif choice == '2' :
        print(data.tail())

    elif choice == '3' :
        print(data.columns)

    elif choice == '4' :
        print(data.dtypes)

    elif choice == '5' :
        print(data.info())

    elif choice == '6' :
        print(data.describe())

    else :
        print("Enter A Valid Choice!")

except :
    print("Something Went Wrong!")
```

```python
print("========== CLEAN DATASET ==========")
print("1. Check Missing Values")
print("2. Handle Missing Values")
print("3. Remove Duplicate Rows")
print("4. Convert Data Types")
print("5. Check Invalid Values")
try:
    choice = input("Enter Your Choice: ")

    if choice == "1":
        print(data.isnull().sum())

    elif choice == "2":
        data = data.ffill()
        print("Missing Values Filled!")

    elif choice == "3":
        data.drop_duplicates(inplace=True)
        print("Duplicate Rows Removed!")

    elif choice == "4":
        data["Date"] = pd.to_datetime(data["Date"])
        print(data.dtypes)

    elif choice == "5":
        print(data[data["Close"] <= 0])

    else:
        print("Invalid Choice!")


except :
    print("Somethig Went Wrong !")
```

```python
print("========== DATA ANALYSIS ==========")
print("1. Calculate Daily Returns")
print("2. Calculate Average Stock Price")
print("3. Find Highest and Lowest Price")
print("4. Calculate Stock Volatility")
print("5. Analyze Trading Volume")
print("6. Calculate Moving Average")
print("7. Analyze Daily Price Changes")

choice = input("Enter Your Choice: ")

if choice == "1":
    data["Daily_Return"] = data["Close"].pct_change() * 100
    print(data.reset_index()[["Date", "Close", "Daily_Return"]])

elif choice == "2":
    print("Average Open Price:", data["Open"].mean())
    print("Average High Price:", data["High"].mean())
    print("Average Low Price:", data["Low"].mean())
    print("Average Close Price:", data["Close"].mean())

elif choice == "3":
    print("Highest Price:", data["High"].max())
    print("Lowest Price:", data["Low"].min())

elif choice == "4":
    volatility = data["Close"].pct_change().std() * 100
    print("Stock Volatility:", volatility)

elif choice == "5":
    print("Total Volume:", data["Volume"].sum())
    print("Average Volume:", data["Volume"].mean())
    print("Highest Volume:", data["Volume"].max())

elif choice == "6":
    data["Moving_Average"] = data["Close"].rolling(7).mean()
    print(data.reset_index()[["Date", "Close", "Moving_Average"]])

elif choice == "7":
    data["Price_Change"] = data["Close"].diff()
    print(data.reset_index()[["Date", "Close", "Price_Change"]])

else:
    print("Invalid Choice!")
```

```python
if "Date" in data.columns:
    data["Date"] = pd.to_datetime(data["Date"])
    data = data.set_index("Date")
print("========== DATA VISUALIZATION ==========")
print("1. Stock Price Line Chart")
print("2. Trading Volume Bar Chart")
print("3. Moving Average Chart")
print("4. Daily Returns Histogram")
print("5. Stock Price Area Chart")
print("6. Correlation Heatmap")

choice = input("Enter Your Choice: ").strip()

if data.empty:
    print("Data Loaded First!")

elif choice == '1':
    info = '''
    1. Weekly Chart
    2. Monthly Chart
    3. Yearly Chart
    '''
    print(info)
    ch = input('Select An Option :- ').strip()

    if ch == '1':
        weekly = data.resample('W')["Close"].mean().reset_index()
        sns.lineplot(data=weekly, x="Date", y="Close", errorbar=None)
        plt.title("Stock Price - Week Wise")
        plt.xlabel("Date")
        plt.ylabel("Average Close Price")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()

    elif ch == '2':
        monthly = data.resample("ME")["Close"].mean().reset_index()
        sns.lineplot(data=monthly, x="Date", y="Close", errorbar=None)
        plt.title("Stock Price - Month Wise")
        plt.xlabel("Date")
        plt.ylabel("Average Close Price")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()

    elif ch == '3':
        yearly = data.resample('YE')["Close"].mean().reset_index()
        sns.lineplot(data=yearly, x="Date", y="Close", errorbar=None)
        plt.title("Stock Price - Year Wise")
        plt.xlabel("Year")
        plt.ylabel("Average Close Price")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()

    else:
        print("Invalid Choice")

elif choice == '2':
    info = '''
    1. Weekly Graph
    2. Monthly Graph
    3. Yearly Graph
    '''
    print(info)
    ch = input('Select An Option :- ').strip()

    if ch == '1':
        weekly = data.resample('W')["Volume"].sum().reset_index()
        sns.barplot(data=weekly, x="Date", y="Volume", errorbar=None)
        plt.title("Trading Volume - Week Wise")
        plt.xlabel("Date")
        plt.ylabel("Total Volume")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()

    elif ch == '2':
        monthly = data.resample("ME")["Volume"].sum().reset_index()
        sns.barplot(data=monthly, x="Date", y="Volume", errorbar=None)
        plt.title("Trading Volume - Month Wise")
        plt.xlabel("Date")
        plt.ylabel("Total Volume")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()

    elif ch == '3':
        yearly = data.resample('YE')["Volume"].sum().reset_index()
        sns.barplot(data=yearly, x="Date", y="Volume", errorbar=None)
        plt.title("Trading Volume - Year Wise")
        plt.xlabel("Year")
        plt.ylabel("Total Volume")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()

    else:
        print("Invalid Choice")

elif choice == '3':
    data["Moving_Average"] = data["Close"].rolling(window=7).mean()
 
    plot_data = data.reset_index()

    sns.lineplot(data=plot_data, x="Date", y="Close", label="Stock Price", errorbar=None)
    sns.lineplot(data=plot_data, x="Date", y="Moving_Average", label="7 Day Moving Average", errorbar=None)

    plt.title("Stock Price with Moving Average")
    plt.xlabel("Date")
    plt.ylabel("Close Price")
    plt.xticks(rotation=45)
    plt.legend()
    plt.tight_layout()
    plt.show()

elif choice == '4':  
    data["Daily_Return"] = data["Close"].pct_change() * 100

    data = data.dropna(subset=["Daily_Return"])
    sns.histplot(data=data, x="Daily_Return", bins=30, kde=True)

    plt.title("Daily Returns Histogram")
    plt.xlabel("Daily Return (%)")
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()

elif choice == '5':
    plot_data = data.reset_index()

    plt.figure(figsize=(12, 6))
    plt.fill_between(plot_data["Date"], plot_data["Close"], color="skyblue", alpha=0.4)
    sns.lineplot(data=plot_data, x="Date", y="Close", errorbar=None, color="blue")

    plt.title("Stock Price Area Chart")
    plt.xlabel("Date")
    plt.ylabel("Close Price")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
elif choice == '6': 
    corr = data[["Open", "High", "Low", "Close", "Volume"]].corr()

    sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
    plt.title("Correlation Heatmap")
    plt.tight_layout()
    plt.show()

else:
    print("Invalid Choice! Please select 1-6")
```

```python
print("========== ANALYSIS RESULTS ==========")
print("1. Display Overall Summary")
print("2. Display Stock Performance")
print("3. Display Risk Analysis")
print("4. Display Trend Analysis")
print("5. Display Final Report")

choice = input("Enter Your Choice: ")

if choice == "1":
    print("========== OVERALL SUMMARY ==========")
    print("Total Trading Days:", len(data))
    print("Average Close Price:", data["Close"].mean())
    print("Highest Price:", data["High"].max())
    print("Lowest Price:", data["Low"].min())
    print("Total Trading Volume:", data["Volume"].sum())

elif choice == "2":
    print("========== STOCK PERFORMANCE ==========")
    first_price = data["Close"].iloc[0]
    last_price = data["Close"].iloc[-1]

    change = last_price - first_price
    percentage = (change / first_price) * 100

    print("Starting Price:", first_price)
    print("Latest Price:", last_price)
    print("Price Change:", change)
    print("Performance:", percentage, "%")

elif choice == "3":
    print("========== RISK ANALYSIS ==========")
    daily_returns = data["Close"].pct_change()
    volatility = daily_returns.std() * 100

    print("Stock Volatility:", volatility, "%")
    print("Maximum Daily Loss:", daily_returns.min() * 100, "%")
    print("Maximum Daily Gain:", daily_returns.max() * 100, "%")

elif choice == "4":
    print("========== TREND ANALYSIS ==========")
    data["Moving_Average"] = data["Close"].rolling(7).mean()

    if data["Close"].iloc[-1] > data["Moving_Average"].iloc[-1]:
        print("Stock Trend: Upward")
    else:
        print("Stock Trend: Downward")
 

elif choice == "5":
    print("========== FINAL STOCK REPORT ==========")
    print(data[["Open", "High", "Low", "Close", "Volume"]].describe())
    print("Overall Performance:")

    if data["Close"].iloc[-1] > data["Close"].iloc[0]:
        print("Stock Price Increased During This Period.")
    elif data["Close"].iloc[-1] < data["Close"].iloc[0]:
        print("Stock Price Decreased During This Period.")
    else:
        print("Stock Price Remained Unchanged.")

else:
    print("Invalid Choice!")
```
